# Tokenização Gemini

```python
# Imports
from google import genai
from dotenv import load_dotenv
import os
```

```python
# Carrega as variáveis do arquivo .env
load_dotenv()
API_KEY= os.getenv("GEMINI_API_KEY")

# Cria o Client do Gemini
client = genai.Client(api_key=API_KEY)
```

```python
prompt = """
Crie um roteiro de viagens internacionais.
Eu quero que liste pelo menos 5 lugares com 
dicas de passeios e entretenimento tanto noturno quanto diurno.
Não liste lugares perigosos.
Leve em consideração baixa temporada, baixo orçamento e variedade de atividades.
Quero um parágrafo contando um pouco sobre o local e os motivos de ser uma boa opção turística
"""
```

```python
prompt = """Atue como um professor de Machine Learning.
Ensine do básico ao avançado sobre cada etapa do desenvolvimento
de um projeto de machine learning.
Depois de dois exemplos de código completo para modelo de Regressão linear,
outro de Árvore de Decisão (Classificação) e outro de KNN(Classificação).
Todos os códigos devem estar comentados.
Dois exemplos de códigos para cada um dos modelos listados.
Não esqueça de avaliar a acurácia, precisão e recall de cada modelo.
"""
```

```python
# Interage com o modelo Gemini
# pergunta = input("Pergunte alguma coisa: ")

resposta = client.models.generate_content(
    model="gemini-3.1-flash-lite",
    contents=prompt
)
```

# Analizando a resposta

# Tokens

```python
tk_prompt = dados['usage_metadata']['prompt_token_count']
tk_resp = dados['usage_metadata']['candidates_token_count']
tk_total = dados['usage_metadata']['total_token_count']

print(f"Tokens prompt: {tk_prompt}")
print(f"Tokens resposta: {tk_resp}")
print(f"Tokens totais: {tk_total}")
```
