```python
from gerador import Gerador
import json
```

```python
gerador = Gerador('gemini-3.5-flash')
```

```python
resposta = gerador.gerar_perguntas('O que são LLMs?')
```

```python
resposta.text
```

```python
perguntas = json.loads(resposta.text)
```

```python
for k,v in perguntas.items():
    print(k)    
    print(v)
    print(30*'-')
```

```python
file_name = input('Nome do arquivo: ').replace(' ', '_')
with open(f'{file_name}.json', 'w', encoding='utf-8') as f:
    json.dump(perguntas, f, indent=4, ensure_ascii=False)
```
